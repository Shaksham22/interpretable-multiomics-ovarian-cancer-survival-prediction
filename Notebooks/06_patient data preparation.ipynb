```python
import pandas as pd

# File paths
input_file = "../Datafile Raw/patient data.xlsx"
output_file = "../Datafiles Final/Datafiles Cleaned/processed_patient_data.xlsx"

# Load the patient data
df = pd.read_excel(input_file)

# Define thresholds from 12 to 66 months (inclusive) at 6-month intervals
thresholds = list(range(12, 67, 6))  # [12, 18, 24, 30, 36, 42, 48, 54, 60, 66]

# Ensure "Overall Survival Status" is in a consistent format (strip spaces, convert to uppercase)
df["Overall Survival Status"] = df["Overall Survival Status"].astype(str).str.strip().str.upper()

# Iterate through each row and assign values for each threshold
for t in thresholds:
    new_column = str(t)
    df[new_column] = 0  # Default value is 0

    for i in range(len(df)):
        survival_months = df.loc[i, "Months of disease-specific survival"]
        survival_status = df.loc[i, "Overall Survival Status"]

        if survival_months <= t:
            if survival_status == "1:DECEASED":
                df.at[i, new_column] = 1
            else:
                df.at[i, new_column] = -1

# Select only the desired columns
cols_to_keep = ["Sample ID", "Patient ID", "Months of disease-specific survival"] + [str(t) for t in thresholds]
processed_df = df[cols_to_keep]

# Save the processed DataFrame to a new Excel file
processed_df.to_excel(output_file, index=False)
print("Processed data saved to:", output_file)
```
