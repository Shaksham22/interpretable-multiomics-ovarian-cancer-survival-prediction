```python
import pandas as pd

month = 42

# File paths
file_paths = {
    "log2cna": f"../Datafiles Final/Datafiles Preprocessed/{month} month/log2cna_preprocessed.pkl",
    "methylation": f"../Datafiles Final/Datafiles Preprocessed/{month} month/methylation_preprocessed.pkl",
    "mrnaseqv2": f"../Datafiles Final/Datafiles Preprocessed/{month} month/mrnaseq_preprocessed.pkl",
}

# Function to load and display data
def load_and_display_pkl(file_path, dataset_name):
    df = pd.read_pickle(file_path)
    print(f"\nDisplaying first 5 rows of {dataset_name} for {month}-month threshold (Shape {df.shape}):")
    display(df.head())
    return df
```

```python
# Load datasets
methylation = load_and_display_pkl(file_paths["methylation"], "methylation")
log2cna = load_and_display_pkl(file_paths["log2cna"], "log2cna")
mrnaseqv2 = load_and_display_pkl(file_paths["mrnaseqv2"], "mrnaseqv2")

# Extract patient IDs (index)
methylation_ids = set(methylation.index)
log2cna_ids = set(log2cna.index)
mrnaseq_ids = set(mrnaseqv2.index)

# Compute intersections
common_all = methylation_ids & log2cna_ids & mrnaseq_ids
common_methylation_log2cna = methylation_ids & log2cna_ids
common_methylation_mrna = methylation_ids & mrnaseq_ids
common_log2cna_mrna = log2cna_ids & mrnaseq_ids

# Display results
print(f"Patients with all three data types: {len(common_all)}")
print(f"Patients with Methylation & log2CNA: {len(common_methylation_log2cna)}")
print(f"Patients with Methylation & mRNA: {len(common_methylation_mrna)}")
print(f"Patients with log2CNA & mRNA: {len(common_log2cna_mrna)}")
```
