```python
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
import pandas as pd

# Step 1: Ensure data is numeric (methylation_df is already normalized)
numeric_data = methylation_df.apply(pd.to_numeric, errors='coerce')

# Step 2: Apply PCA
pca = PCA(n_components=2)  # Extract first two principal components
pca_result = pca.fit_transform(numeric_data)

# Step 3: Convert PCA result into a DataFrame while keeping patient IDs
pca_df = pd.DataFrame(pca_result, index=methylation_df.index, columns=['PC1', 'PC2'])

# Step 4: Plot PC1 vs PC2
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA: First Two Principal Components')
plt.grid(True)

# Show the plot
plt.show()
```

```python
from sklearn.cluster import KMeans

# Apply K-Means clustering (choose k based on your data)
kmeans = KMeans(n_clusters=2, random_state=42)  # Adjust 'n_clusters' as needed
pca_df['Cluster'] = kmeans.fit_predict(pca_df)

# Scatter plot with cluster colors
plt.figure(figsize=(8,6))
plt.scatter(pca_df['PC1'], pca_df['PC2'], c=pca_df['Cluster'], cmap='viridis', alpha=0.7)
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA with K-Means Clustering')
plt.grid(True)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Apply PCA
pca = PCA()
pca_result = pca.fit_transform(methylation_df)

# Explained variance ratio
explained_variance = np.cumsum(pca.explained_variance_ratio_)

# Plot cumulative explained variance
plt.figure(figsize=(8,6))
plt.plot(range(1, len(explained_variance) + 1), explained_variance, marker='o', linestyle='--')
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Explained Variance vs. Number of Principal Components")
plt.axhline(y=0.95, color='r', linestyle='--', label='95% Variance Threshold')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

# Define features (X) and target labels (y)
X = methylation_df.drop(columns=['Target'])  # Assuming 'Target' is the patient condition
y = methylation_df['Target']

# Select top 500 genes
selector = SelectKBest(score_func=f_classif, k=500)
X_selected = selector.fit_transform(X, y)

# Get selected feature names
selected_features = X.columns[selector.get_support()]
print("Top Selected Features:", selected_features)
```
