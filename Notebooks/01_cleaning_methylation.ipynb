```python
import pandas as pd
from sklearn.impute import KNNImputer
import matplotlib.pyplot as plt
import seaborn as sns


# Define the file path
file_path = '../Datafile Raw/ov_tcga_pan_can_atlas_2018/data_methylation_hm27_hm450_merged.txt'
# Load the data
df = pd.read_csv(file_path, sep="\t", low_memory=False)
df.head()


```

```python
df=df.transpose()
df
```

```python
patient_data = df.copy()  
patient_data = patient_data.drop(index=df.index[[ 2, 3]])
patient_data = patient_data.drop(columns=[patient_data.columns[18351]])


```

```python
# Define the threshold for missing data (20%)
missing_threshold = 0.1

# Calculate the percentage of missing values for each patient (row)
missing_percentage_per_patient = patient_data.isna().mean(axis=1)

# Get the index of patients with more than 20% missing data
patients_with_high_missing = missing_percentage_per_patient[missing_percentage_per_patient > missing_threshold].index.tolist()

# Print the results
print(f"Total number of patients with more than {missing_threshold*100}% missing data: {len(patients_with_high_missing)}")
print(f"Patients with more than {missing_threshold*100}% or more missing data:")
print(patients_with_high_missing)
```

```python
# Define the threshold for missing data (80%)
missing_threshold = 0.35

# Calculate the percentage of missing values for each column
missing_percentage_per_column = patient_data.isna().mean()

# Filter columns where missing percentage is ≥ 80%
high_missing_sites = missing_percentage_per_column[missing_percentage_per_column >= missing_threshold].index.tolist()

# Print the CpG sites with 80% or more missing data
print(f"Total CpG sites with at least {missing_threshold*100}% missing data: {len(high_missing_sites)}")
print(f"CpG sites with {missing_threshold*100}% or more missing data:")
print(high_missing_sites)
```

```python


# Ensure the first two rows (Stable ID & Gene Name) remain unchanged
stable_id_row = patient_data.iloc[0]  # First row (Stable ID)
gene_name_row = patient_data.iloc[1]  # Second row (Gene Name)

# Extract only patient data (3rd row onwards)
patient_values = patient_data.iloc[2:].copy()

# Initialize KNN Imputer
knn_imputer = KNNImputer(n_neighbors=5)  # You can adjust n_neighbors

# Impute each column separately
for col in patient_values.columns:
    patient_values[[col]] = knn_imputer.fit_transform(patient_values[[col]])

# Reconstruct the full DataFrame
patient_data = pd.concat([stable_id_row.to_frame().T, gene_name_row.to_frame().T, patient_values])

```

```python
# Check for NaN values in the first row (Stable ID) and second row (Gene Name)
first_row_nan = patient_data.iloc[0].isna().sum()
second_row_nan = patient_data.iloc[1].isna().sum()

# Print results
print(f"🔍 Missing values in the first row (Stable ID): {first_row_nan}")
print(f"🔍 Missing values in the second row (Gene Name): {second_row_nan}")

# Check if any missing values exist
if first_row_nan == 0 and second_row_nan == 0:
    print("✅ No missing values found in the first and second row.")
else:
    print("⚠️ There are missing values in the first or second row.")
```

```python
nan_count = patient_data.isna().sum().sum()

# Print results
if nan_count > 0:
    print(f"The methylation dataset contains {nan_count} missing (NaN) values.")
else:
    print("No missing (NaN) values found in the methylation dataset.")
```
